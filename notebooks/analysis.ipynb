```python
from utils.analysis import plot_comparison_bars, plot_paired_delta

common_training_params = {'chn_depth': 2, 'tem_depth': 2, 'spec_depth': 2, 'reconstruction_lambda':0.001}
common_training_params.update()

Chn_attn = common_training_params | {
    'chn_attn_flag': True,
    'spectrum_attn_flag': False,
    'temporal_attn_flag': False
}
ChnSpectrum_attn = common_training_params | {
    'chn_attn_flag': True,
    'spectrum_attn_flag': True,
    'temporal_attn_flag': False
}
ChnTemporal_attn = common_training_params | {
    'chn_attn_flag': True,
    'spectrum_attn_flag': False,
    'temporal_attn_flag': True
}
ChnSpectrumTemporal_attn = common_training_params | {
    'chn_attn_flag': True,
    'spectrum_attn_flag': True,
    'temporal_attn_flag': True
}

configs = {
    'mtf_c (chn_attn)': ({'n_repeats': 1}, Chn_attn, 'mtf_c'),
    'mtf_c (chn+spectrum_attn)': ({'n_repeats': 1}, ChnSpectrum_attn, 'mtf_c'),
    'mtf_c (chn+temporal_attn)': ({'n_repeats': 1}, ChnTemporal_attn, 'mtf_c'),
    'mtf_c (chn+spectrum+temporal_attn)': ({'n_repeats': 1}, ChnSpectrumTemporal_attn, 'mtf_c'),
}

plot_comparison_bars(
    config_pairs=configs,
    root_dir='experiments/classification/frequency_backbone',
    save_path='plots/classification/frequency_backbone',
    filename='mtfc_branch_attention_ablation',
    figsize=(14, 8),
    metric='mean_acc',
    error_metric='std_acc'
)

configs = {
    'mtf_c (chn+spectrum_attn)':{
        'baseline': ({'n_repeats': 1}, Chn_attn, 'mtf_c'),
        'experiment': ({'n_repeats': 1}, ChnSpectrum_attn, 'mtf_c'),
    },
    'mtf_c (chn+temporal_attn)':{
        'baseline': ({'n_repeats': 1}, Chn_attn, 'mtf_c'),
        'experiment': ({'n_repeats': 1}, ChnTemporal_attn, 'mtf_c'),
    },
    'mtf_c (chn+spectrum+temporal_attn)':{
        'baseline': ({'n_repeats': 1}, Chn_attn, 'mtf_c'),
        'experiment': ({'n_repeats': 1}, ChnSpectrumTemporal_attn, 'mtf_c'),
    },
}

plot_paired_delta(
    config_pairs=configs,
    reference_key='baseline',
    root_dir='experiments/classification/frequency_backbone',
    save_path='plots/classification/frequency_backbone',
    filename='mtfc_branch_attention_comparison',
    figsize=(12, 6),
    metric='mean_acc',
    hline=None
)

clear_output()
```
